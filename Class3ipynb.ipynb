```python
class Agent:
  def __init__(self,agent_id,agent_type):
    self.agent_id=agent_id
    self.agent_type=agent_type
    self.utility=0

  def introduce(self):
    print(f"I am (self.agent.id) and my type is (self.agent_type)")
    print(f"My utility is (self.utility)")
a1=Agent(1,0)
```

```python
print(a1)
print(a1.agent_id)
print(a1.utility)
```

```python
a2=Agent("Max","cooperator")
print(a2)
print(a2.agent_id)
print(a2.utility)
```

```python
a2.introduce()
```

```python
import numpy as np
N=20
probe=0.2
# generate 20 ids
ids=list(range(100,100+N))
print(ids)
types=np.random.binomial(1,probe,N)
print(types)

#list of agents
A=[]
for i in range(N):
  A.append(Agent(ids[i],types[i]))
```

```python
print(A)
print(len(A))
for i in range (N):
  A[i].introduce()
```

```python
for i in A:
  i.introduce()
```

```python
class Payoffs:
  def __init__(self,payoff_matrix):
    self.pm=payoff_matrix

  def play(self,p1,p2): #p1 and p2 - actions (0 or 1 if two strategies)
    return self.pm[p1,p2]

  def play2(self,p1,p2):
    return self.pm[p1,p2],self.pm[p2,p1]

PD=Payoffs(np.array([(4,0),[5,1]])) #Prisoner's Dilemma Game
print(PD.pm)

choice1=0
choice2=0
print(PD.play(choice1,choice2))
print(PD.play(choice2.choice1))

print(PD.play2(choice1,choice2))
```
