```python
import numpy as np
class PayoffMatrix:
  def __init__(self, p1,p2,p3,p4,a1,a2,a3,a4):
    self.p1=p1
    self.p2=p2
    self.p3=p3
    self.p4=p4

    self.a1=a1
    self.a2=a2
    self.a3=a3
    self.a4=a4

  class Payoffs:
    def __init__(self, payoff_matrix):
        self.PM = payoff_matrix

    def play (self,r,c):
      return self.PM[r,c]
```

```python
class Agent:
    def __init__(self, agent_id, agent_type):
       self.agent_id = agent_id
       self.agent_type = agent_type
       self.utility = 0

    def introduce(self):
        print(f"I am {self.agent_id} and my type is {self.agent_type}")
        print(f"My utility is {self.utility}")

a1 = Agent(1, 0)
print(a1)
print(a1.agent_id)
print(a1.agent_type)
print(a1.utility)
```

```python
N = 10
prob0 = 0.5
# generate N ids
ids = list(range(200, 200 + N))
print(ids)
types = np.random.binomial(1, prob0, N)
print(types)

#list of agents
A = []
for i in range(N):
    A.append(Agent(ids[i], types[i]))
```

```python
for a in A:
  a.introduce()
```

```python
for a in A:
  a.utility=1
for a in A:
    a.introduce()
```

```python
M=np.array([[4,1],[3,-4]])
game=PayoffMatrix.Payoffs(M)
print("row=0,col=0:",game.play(0,0)) # 0 means vaccinated, 1 means non-vaccinated
print("row=0,col=1:",game.play(0,1))
print("row=1,col=0:",game.play(1,0))
print("row=1,col=1:",game.play(1,1))
```

```python
for a in A:
  col=a.agent_type
  a.utility=game.play(0,col)
for a in A:
    print(f"type={a.agent_type} utility={a.utility}")
```

```python
for a in A:
  col=a.agent_type
  a.utility=game.play(1,col)
for a in A:
    print(f"type={a.agent_type} utility={a.utility}")
```
